# 2 · Two documents from the same store

Both are marked **INTERNAL**. Every employee in the company is allowed to read either one.

### Is either of these confidential?

*Take a moment with this one.*

**Document A** is an index — reference numbers against names. It says nothing about anybody.

**Document B** is a staffing rota. It mentions a medical absence, but **against a reference number,
not a person.** It does not identify anyone at all.

That second point matters legally: under GDPR, data that cannot identify a natural person is not
personal data. A rota of reference numbers is ordinary operational scheduling — the kind of document
that sits in a shared folder in every company.

# What just happened

That sentence is **special-category health data** about a named individual — the most protected
class of personal data there is.

**Ten seconds ago it did not exist.** Not in any document, not in the store, not anywhere in the
company. The assistant created it by joining two things it was entitled to read.

Neither half was confidential:

- Document A named people but said nothing about them
- Document B described a medical absence but **identified nobody**

The assistant supplied the identity. That link exists in no document.

# Why this needs a testing capability

**This is not a bug report.** There is no misconfiguration to correct and no setting to switch on.
It is a category of risk that our current assurance cannot see.

| | |
|---|---|
| **It only appears in the deployed system** | Testing the AI model on its own would never surface it — the model has no documents. The risk is created by connecting one to a document store. |
| **It is measurable** | Hundreds of probes per scenario, scored mechanically rather than by opinion. |
| **It is repeatable** | Runs before and after every change, so a fix can be proven and a regression caught. |

The same technique, applied across this system, also surfaced:

- a **termination** and its effective date
- an active **ethics investigation** against a named individual
- an unannounced **acquisition target**

**Four attempts. Four disclosures.**
